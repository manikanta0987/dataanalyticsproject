```python
leakage_columns = [
    column for column in df.columns
    if "Naive_Bayes_Classifier" in column
]

df.drop(columns=leakage_columns, inplace=True)
df
```

```python
df.columns = df.columns.str.strip()
df
```

```python
df["Education_Level"] = df["Education_Level"].fillna("Unknown")
df["Marital_Status"] = df["Marital_Status"].fillna("Unknown")
df["Income_Category"] = df["Income_Category"].fillna("Unknown")
df
```

```python
df["Churn_Flag"] = (
    df["Attrition_Flag"]
    .eq("Attrited Customer")
    .astype("int8")
)
df
```

```python
df["Age_Group"] = pd.cut(
    df["Customer_Age"],
    bins=[0, 30, 40, 50, 60, 100],
    labels=[
        "Under 30",
        "30-39",
        "40-49",
        "50-59",
        "60+"
    ],
    right=False
)
df
```

```python
df["Tenure_Years"] = (
    df["Months_on_book"] / 12
).round(1)
df
```

```python
df["Avg_Transaction_Value"] = np.where(
    df["Total_Trans_Ct"] > 0,
    df["Total_Trans_Amt"] / df["Total_Trans_Ct"],
    0
)
df
```

```python
df["Utilization_Band"] = pd.cut(
    df["Avg_Utilization_Ratio"],
    bins=[-0.001, 0.25, 0.50, 0.75, 1.00],
    labels=["Very Low", "Low", "Medium", "High"],
    include_lowest=True
)
df
```

```python
df.to_csv("BankChurners_Cleaned.csv", index=False)
```

```python
df.groupby("Attrition_Flag").agg(
    avg_transactions=("Total_Trans_Ct", "mean"),
    avg_transaction_amount=("Total_Trans_Amt", "mean"),
    avg_inactive_months=("Months_Inactive_12_mon", "mean"),
    avg_contacts=("Contacts_Count_12_mon", "mean"),
    avg_products=("Total_Relationship_Count", "mean"),
    avg_utilization=("Avg_Utilization_Ratio", "mean")
).round(2)
```

```python
columns_to_analyze = [
    "Gender",
    "Age_Group",
    "Income_Category",
    "Education_Level",
    "Marital_Status",
    "Card_Category",
    "Months_Inactive_12_mon",
    "Contacts_Count_12_mon",
    "Total_Relationship_Count",
    "Utilization_Band"
]

churn_analysis_results = {}

for column in columns_to_analyze:

    result = (
        df.groupby(
            column,
            dropna=False,
            observed=True
        )
        .agg(
            total_customers=("CLIENTNUM", "nunique"),
            churned_customers=("Churn_Flag", "sum"),
            churn_rate=("Churn_Flag", "mean")
        )
        .reset_index()
    )

    result["churn_rate"] = (
        result["churn_rate"] * 100
    ).round(2)

    churn_analysis_results[column] = result
    
```

```python
for column, result in churn_analysis_results.items():
    print(f"Churn Analysis by {column}")
    display(result)
```
